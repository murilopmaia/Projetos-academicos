## Eliminação Gaussiana com Substituição Regressiva

O algoritmo fundamental que transforma o sistema linear em uma matriz triangular superior para depois resolver as incógnitas de trás para frente.

```python
def gaussian_elimination_backward_substitution(n, A):
    """
    Resolve o sistema linear nxn usando Eliminação de Gauss com Substituição Reversa.
    :param n: número de equações e incógnitas.
    :param A: matriz aumentada [A | b] de dimensão n x (n+1).
    :return: lista com a solução [x_1, ..., x_n] ou None.
    """
    
    # ==========================================
    # FASE 1: ELIMINAÇÃO (Transformando em Triangular Superior)
    # ==========================================
    
    # Passo 1: Percorre as colunas da matriz até a penúltima (n-1).
    # O objetivo é "zerar" tudo que está abaixo da diagonal principal nesta coluna 'i'.
    for i in range(n - 1):
        
        # Passo 2: Busca por um pivô válido.
        # Ele desce pela coluna 'i' procurando o primeiro número que não seja zero.
        # Isso é crucial porque não podemos dividir por zero na hora de calcular o multiplicador.
        p = -1
        for k in range(i, n):
            if A[k][i] != 0:
                p = k  # Encontrou uma linha com um elemento não nulo!
                break

        # Se não achou nenhum número diferente de zero na coluna inteira, 
        # o sistema não tem solução única (a matriz é singular).
        if p == -1:
            print("no unique solution exists")
            return None

        # Passo 3: Troca de linhas (Pivoteamento trivial)
        # Se o primeiro número não nulo que achamos (linha 'p') não está na posição
        # do pivô atual (linha 'i'), nós invertemos as duas linhas inteiras de lugar.
        if p != i:
            A[p], A[i] = A[i], A[p]

        # Passo 4: Começa a zerar as linhas abaixo do pivô 'i'.
        # 'j' representa as linhas que estão abaixo da linha do pivô atual.
        for j in range(i + 1, n):
            
            # Passo 5: Calcula o multiplicador (m).
            # É a razão entre o elemento que queremos zerar e o pivô.
            m_ji = A[j][i] / A[i][i]
            
            # Passo 6: Atualiza a linha inteira 'j' (incluindo o termo independente 'b' no final).
            # Nova linha j = (Linha j atual) - m * (Linha do pivô).
            for k in range(n + 1):
                A[j][k] = A[j][k] - m_ji * A[i][k]


    # ==========================================
    # FASE 2: SUBSTITUIÇÃO REGRESSIVA (Resolvendo de baixo para cima)
    # ==========================================
    
    # Passo 7: Verificação de segurança final.
    # Se o último elemento da diagonal principal ficou zero, impossível resolver a última variável.
    if A[n - 1][n - 1] == 0:
        print("no unique solution exists")
        return None

    # Passo 8: Cria uma lista vazia para guardar as respostas (nossos x's).
    x = [0.0] * n
    # Resolve a última equação diretamente: isola o x_n passando o coeficiente dividindo.
    x[n - 1] = A[n - 1][n] / A[n - 1][n - 1]

    # Passo 9: "Sobe" a escada, resolvendo do penúltimo x até o primeiro (índice 0).
    for i in range(n - 2, -1, -1):
        # Calcula a soma dos produtos dos coeficientes pelas variáveis que JÁ descobrimos.
        soma = sum(A[i][j] * x[j] for j in range(i + 1, n))
        
        # Isola a variável atual (x_i), subtraindo a soma do termo independente e dividindo pelo coeficiente da diagonal.
        x[i] = (A[i][n] - soma) / A[i][i]

    # Passo 10: Retorna o vetor com todas as variáveis calculadas!
    return x
```

### Resolvendo o sistema classico

```python
n = len(b_classic)
A_aug = [A_classic[i] + [b_classic[i]] for i in range(n)]

x_calc = gaussian_elimination_backward_substitution(n, A_aug)

print("--- RESULTADOS DA RESOLUÇÃO ---")
print(f"Vetor x calculado: {x_calc}\n")

b_calc = [0.0] * n
for i in range(n):
    soma = sum(A_classic[i][j] * x_calc[j] for j in range(n))
    b_calc[i] = soma

print("--- COMPROVAÇÃO (A * x_calculado) ---")
print(f"b calculado: {b_calc}")
print(f"b original:  {b_classic}\n")

erros_absolutos = [abs(b_classic[i] - b_calc[i]) for i in range(n)]
erro_maximo = max(erros_absolutos)

print("--- ANÁLISE DE ERRO ---")
for i in range(n):
    print(f"Erro na equação {i+1}: {erros_absolutos[i]:.4e}")
print(f"\nErro Máximo (Norma Infinita): {erro_maximo:.4e}")
```

### Vantagens e Desvantagens



#### Vantagens

- **Método geral:** Pode ser aplicado à resolução de qualquer sistema linear quadrado, desde que exista uma solução única.

- **Alta precisão:** Quando combinado com pivotamento parcial, o método apresenta boa estabilidade numérica para a maioria dos problemas práticos.

- **Algoritmo determinístico:** A solução é obtida por meio de uma sequência bem definida de operações elementares, garantindo resultados reprodutíveis.

#### Desvantagens

- **Alto custo computacional:** O método possui complexidade de **O(n³)**, tornando-se menos eficiente para sistemas lineares de grande porte.

- **Maior consumo de memória:** Para matrizes densas, é necessário armazenar todos os coeficientes da matriz durante o processo de eliminação.

- **Pouco eficiente para matrizes esparsas:** O processo de eliminação pode gerar novos elementos não nulos (fenômeno conhecido como *fill-in*), aumentando o tempo de processamento e o consumo de memória.

- **Menor desempenho em estruturas especiais:** Para matrizes com propriedades específicas, como matrizes tridiagonais, existem algoritmos especializados, como o método de Thomas ou a fatoração de Crout para matrizes tridiagonais, que apresentam menor custo computacional e maior eficiência.

## Metodos da Elminiçao de Gauss com Pivotamento Parcial 


O Método da Eliminação de Gauss com Pivotamento Parcial é uma melhoria da Eliminação de Gauss tradicional. Seu principal objetivo é aumentar a estabilidade numérica do algoritmo, reduzindo os erros de arredondamento que podem ocorrer durante o processo de eliminação.

```python
def gaussian_elimination_partial_pivoting(n, A):
    """
    Resolve o sistema linear nxn usando Eliminação de Gauss com Pivoteamento Parcial.
    :param n: número de equações e incógnitas.
    :param A: matriz aumentada [A | b] de dimensão n x (n+1).
    :return: lista com a solução [x_1, ..., x_n] ou None.
    """
    
    # Passo 1: Inicializa o vetor de ponteiros de linha.
    # NROW = [0, 1, 2, ..., n-1]. Ele guarda a ordem "virtual" das linhas.
    # Em vez de movermos a linha A[0], moveremos NROW[0].
    NROW = [i for i in range(n)]

    # ==========================================
    # FASE 1: ELIMINAÇÃO COM PIVOTEAMENTO
    # ==========================================
    
    # Passo 2: Percorre as colunas para fazer a eliminação.
    for i in range(n - 1):
        
        # Passo 3: A Caçada pelo Maior Pivô!
        # Assume inicialmente que o maior valor está na diagonal (posição i).
        p = i
        max_val = abs(A[NROW[i]][i]) 
        
        # Procura nas linhas abaixo se existe algum número maior (em módulo).
        for j in range(i + 1, n):
            if abs(A[NROW[j]][i]) > max_val:
                max_val = abs(A[NROW[j]][i])
                p = j # Guarda o índice da linha que tem o novo maior valor.

        # Passo 4: Se o maior valor encontrado for zero, a coluna inteira é zero.
        # O sistema não tem solução única.
        if A[NROW[p]][i] == 0:
            print("no unique solution exists")
            return None

        # Passo 5: A Troca Virtual de Linhas (Simulated row interchange).
        # Se o maior valor não está na linha atual 'i', nós apenas trocamos 
        # os índices no vetor NROW. A matriz 'A' na memória fica intacta!
        if NROW[i] != NROW[p]:
            NCOPY = NROW[i]
            NROW[i] = NROW[p]
            NROW[p] = NCOPY

        # Passo 6: Começa a eliminação propriamente dita.
        for j in range(i + 1, n):
            # Passo 7: Calcula o multiplicador. 
            # Note que acessamos A usando A[NROW[...]] para respeitar a nova ordem virtual.
            # Graças ao pivoteamento, garantimos que abs(m) <= 1.
            m = A[NROW[j]][i] / A[NROW[i]][i]
            
            # Passo 8: Atualiza a linha 'j' subtraindo a linha do pivô multiplicada por 'm'.
            for k in range(n + 1):
                A[NROW[j]][k] = A[NROW[j]][k] - m * A[NROW[i]][k]

    # ==========================================
    # FASE 2: SUBSTITUIÇÃO REGRESSIVA
    # ==========================================
    
    # Passo 9: Checa se o último pivô (virtual) é zero.
    if A[NROW[n - 1]][n - 1] == 0:
        print("no unique solution exists")
        return None

    # Passo 10: Prepara o vetor de solução e resolve a última variável (x_n).
    x = [0.0] * n
    x[n - 1] = A[NROW[n - 1]][n] / A[NROW[n - 1]][n - 1]

    # Passo 11: Sobe a escada resolvendo o resto do sistema.
    # A lógica é idêntica à eliminação básica, mas sempre consultando NROW
    # para saber qual é a "verdadeira" linha que devemos ler.
    for i in range(n - 2, -1, -1):
        soma = sum(A[NROW[i]][j] * x[j] for j in range(i + 1, n))
        x[i] = (A[NROW[i]][n] - soma) / A[NROW[i]][i]

    # Passo 12: Retorna a solução!
    return x
```

## Método da Eliminação de Gauss com Pivotamento Escalado


E uma variação da Eliminação de Gauss que busca aumentar a estabilidade numérica durante o processo de eliminação. Diferentemente do pivotamento parcial, a escolha do pivô não é baseada apenas no maior elemento da coluna, mas na razão entre o elemento candidato e o maior elemento da sua própria linha.

Esse procedimento evita que linhas com coeficientes muito grandes dominem a escolha do pivô, reduzindo os efeitos de erros de arredondamento em sistemas mal condicionados.

```python
def gaussian_elimination_scaled_partial_pivoting(n, A):
    """
    Resolve o sistema linear nxn usando Eliminação de Gauss com Pivoteamento Parcial Escalado.
    :param n: número de equações e incógnitas.
    :param A: matriz aumentada [A | b] de dimensão n x (n+1).
    """
    # 's' é o vetor que vai guardar o Fator de Escala de cada linha.
    s = [0.0] * n
    # 'NROW' continua sendo nosso vetor de troca virtual de linhas.
    NROW = [0] * n
    
    # ==========================================
    # PASSO PRELIMINAR: CALCULANDO AS ESCALAS
    # ==========================================
    
    # Step 1: Varre todas as linhas para preencher 's' e 'NROW'.
    for i in range(n):
        # Encontra o maior elemento (em módulo) APENAS dos coeficientes da matriz A (j de 0 até n-1).
        s[i] = max(abs(A[i][j]) for j in range(n)) 
        
        # Se o maior elemento de uma linha for zero, significa que a linha toda é de zeros.
        if s[i] == 0:
            print("no unique solution exists")
            return None
        
        NROW[i] = i
        
    # ==========================================
    # FASE 1: ELIMINAÇÃO COM PIVOTEAMENTO ESCALADO
    # ==========================================
    
    # Step 2: Percorre as colunas para fazer a eliminação.
    for i in range(n - 1):
        
        # Step 3: A nova caçada pelo pivô usando o tamanho RELATIVO.
        p = i
        # Em vez de olhar só o valor, dividimos pelo fator de escala da linha correspondente.
        max_razao = abs(A[NROW[i]][i]) / s[NROW[i]]
        
        # Procura nas linhas abaixo se alguém tem uma razão maior.
        for j in range(i + 1, n):
            razao_atual = abs(A[NROW[j]][i]) / s[NROW[j]]
            if razao_atual > max_razao:
                max_razao = razao_atual
                p = j # Guarda a linha vencedora.
        
        # Step 4: Se até o melhor pivô for zero, a coluna é nula.
        if A[NROW[p]][i] == 0:
            print("no unique solution exists")
            return None
            
        # Step 5: A troca virtual de linhas (idêntica ao método anterior).
        if NROW[i] != NROW[p]:
            NCOPY = NROW[i]
            NROW[i] = NROW[p]
            NROW[p] = NCOPY
            
        # Step 6: Eliminação propriamente dita (idêntica ao método anterior).
        for j in range(i + 1, n):
            # Step 7: Multiplicador.
            m = A[NROW[j]][i] / A[NROW[i]][i]
            # Step 8: Atualiza a linha.
            for k in range(n + 1):
                A[NROW[j]][k] = A[NROW[j]][k] - m * A[NROW[i]][k]
                
    # ==========================================
    # FASE 2: SUBSTITUIÇÃO REGRESSIVA (Inalterada)
    # ==========================================
    
    # Step 9
    if A[NROW[n - 1]][n - 1] == 0:
        print("no unique solution exists")
        return None
        
    # Step 10
    x = [0.0] * n
    x[n - 1] = A[NROW[n - 1]][n] / A[NROW[n - 1]][n - 1]
    
    # Step 11
    for i in range(n - 2, -1, -1):
        soma = sum(A[NROW[i]][j] * x[j] for j in range(i + 1, n))
        x[i] = (A[NROW[i]][n] - soma) / A[NROW[i]][i]
        
    # Step 12
    return x
```

## Método da Fatoração LU


consiste em decompor uma matriz quadrada A no produto de duas matrizes triangulares:

$$
A=LU
$$

onde:

* L é uma matriz triangular inferior com diagonal principal composta por 1 (na fatoração de Doolittle);
* U é uma matriz triangular superior.

Após realizar a fatoração, a resolução do sistema linear

```python
def lu_factorization(n, A):
    """
    Fatora a matriz A nxn no produto LU.
    :param n: dimensão da matriz.
    :param A: matriz de coeficientes n x n.
    """
    # Inicializa as matrizes L e U com zeros
    L = [[0.0] * n for _ in range(n)]
    U = [[0.0] * n for _ in range(n)]
    
    # Por definição (Fatoração de Doolittle), a diagonal principal de L é sempre composta por 1s.
    for i in range(n):
        L[i][i] = 1.0
        
    # Step 1: O primeiro elemento de U é simplesmente o primeiro elemento de A.
    U[0][0] = A[0][0] / L[0][0] # Como L[0][0] é 1, U[0][0] = A[0][0]
    
    # Se o primeiro pivô for zero, a fatoração falha (sem pivoteamento).
    if L[0][0] * U[0][0] == 0:
        print("Factorization impossible")
        return None, None
        
    # Step 2: Calcula a primeira linha de U e a primeira coluna de L.
    for j in range(1, n):
        U[0][j] = A[0][j] / L[0][0]  # Copia a primeira linha de A para U
        L[j][0] = A[j][0] / U[0][0]  # Calcula os primeiros multiplicadores de L
        
    # Step 3: O coração do algoritmo. Percorre o miolo da matriz (da linha/coluna 1 até n-2).
    for i in range(1, n - 1):
        
        # Step 4: Calcula o elemento da diagonal principal de U.
        # Subtrai do elemento original de A o produto dos elementos já calculados em L e U.
        soma_diag = sum(L[i][k] * U[k][i] for k in range(i))
        U[i][i] = A[i][i] - soma_diag
        
        if L[i][i] * U[i][i] == 0:
            print("Factorization impossible")
            return None, None
            
        # Step 5: Calcula o resto da linha i de U e da coluna i de L.
        for j in range(i + 1, n):
            # Preenchendo a linha 'i' de U (calculando da esquerda para a direita)
            soma_u = sum(L[i][k] * U[k][j] for k in range(i))
            U[i][j] = (1.0 / L[i][i]) * (A[i][j] - soma_u)
            
            # Preenchendo a coluna 'i' de L (calculando de cima para baixo - os multiplicadores)
            soma_l = sum(L[j][k] * U[k][i] for k in range(i))
            L[j][i] = (1.0 / U[i][i]) * (A[j][i] - soma_l)
            
    # Step 6: Calcula o último elemento da matriz, na ponta inferior direita de U.
    soma_nn = sum(L[n - 1][k] * U[k][n - 1] for k in range(n - 1))
    U[n - 1][n - 1] = A[n - 1][n - 1] - soma_nn
    
    # Step 7: Devolve as duas matrizes prontas!
    return L, U

def solve_lu(n, L, U, b):
    """
    Resolve o sistema linear LUx = b.
    """
    # Passo 1: Resolver Ly = b (Substituição Progressiva / De cima para baixo)
    y = [0.0] * n
    for i in range(n):
        # A lógica é isolar y[i]. Como a matriz L é triangular inferior,
        # só precisamos do que já foi calculado nas linhas de cima.
        soma = sum(L[i][j] * y[j] for j in range(i))
        y[i] = (b[i] - soma) / L[i][i]
        
    # Passo 2: Resolver Ux = y (Substituição Regressiva / De baixo para cima)
    x = [0.0] * n
    for i in range(n - 1, -1, -1):
        # Idêntico ao final da Eliminação de Gauss normal.
        soma = sum(U[i][j] * x[j] for j in range(i + 1, n))
        x[i] = (y[i] - soma) / U[i][i]
        
    return x

def print_matrix(name, matrix):
    print(f"Matriz {name}:")
    for row in matrix:
        print("  [" + ", ".join(f"{val:8.4f}" for val in row) + "]")
    print()
```

```python
n = len(b_classic)

# --- FATORAÇÃO
print("--- 1. REALIZANDO FATORAÇÃO LU... ---")
L, U = lu_factorization(n, A_classic)
print_matrix("L", L)
print_matrix("U", U)

# --- Resolver para b1
print(f"--- 2. RESOLVENDO PARA b1 = {b_classic} ---")
x1 = solve_lu(n, L, U, b_classic)
print(f"Vetor x1 encontrado: {x1}\n")

# --- Resolver para b2
print(f"--- 3. RESOLVENDO PARA b2 = {b2_classic} ---")
x2 = solve_lu(n, L, U, b2_classic)
print(f"Vetor x2 encontrado: {x2}\n")

# --- ETAPA 4: COMPROVAÇÃO PARA O NOVO SISTEMA ---
b_novo_calc = [0.0] * n
for i in range(n):
    soma = sum(A_classic[i][j] * x2[j] for j in range(n))
    b_novo_calc[i] = soma

print("--- COMPROVAÇÃO (A * x2) ---")
print(f"b2 calculado: {b_novo_calc}")
print(f"b2 esperado:  {b2_classic}")

erros_absolutos = [abs(b2_classic[i] - b_novo_calc[i]) for i in range(n)]
print(f"Erro Máximo (Norma Infinita): {max(erros_absolutos):.4e}")
```

#### Vantagens
* Eficiência para múltiplos sistemas: Após calcular a fatoração A=LU, diferentes sistemas com a mesma matriz A e vetores b distintos podem ser resolvidos rapidamente utilizando apenas substituições direta e reversa.
* Menor custo em problemas repetitivos: A fatoração é realizada apenas uma vez, reduzindo significativamente o tempo computacional quando vários sistemas compartilham a mesma matriz de coeficientes.
* Método amplamente utilizado: A fatoração LU é um dos algoritmos fundamentais da Álgebra Linear Numérica e serve como base para diversos softwares científicos.
* Facilidade de implementação: O algoritmo de Doolittle é relativamente simples e organiza naturalmente o processo de decomposição em duas matrizes triangulares.
* Boa eficiência computacional: A fatoração possui complexidade O(n3) e cada resolução posterior do sistema possui complexidade O(n2).

#### Desvantagens
* Necessidade de pivôs não nulos: A fatoração LU sem pivotamento pode falhar quando ocorre um pivô nulo ou muito pequeno, exigindo o uso de pivotamento parcial.
* Maior consumo de memória: É necessário armazenar simultaneamente as matrizes L e U.
* Não é o método mais eficiente para matrizes especiais: Para matrizes simétricas definidas positivas ou tridiagonais, métodos como Cholesky, LDLᵀ ou Crout apresentam menor custo computacional.
* Sensibilidade numérica: Sem pivotamento, o método pode produzir erros significativos em sistemas mal condicionados.
* Maior custo para um único sistema: Quando apenas um sistema linear precisa ser resolvido, a fatoração LU pode não oferecer vantagens significativas em relação à Eliminação de Gauss tradicional, pois ambas possuem a mesma ordem de complexidade computacional.

## Factoraçao $LDL^T$


O método da Fatoração $LDL^T$ é uma técnica de decomposição matricial utilizada para fatorar uma matriz simétrica e definida positiva em três matrizes mais simples.
Onde: 
* L é uma matriz triangular inferior com elementos iguais a 1 na diagonal principal;
* D é uma matriz diagonal;
* $L^T$ é a transposta de L.

Ao contrário da fatoração de Cholesky, a fatoração LDLᵀ não utiliza raízes quadradas, tornando-a computacionalmente mais eficiente e numericamente estável em diversas aplicações.

```python
def ldlt_factorization(n, A):
    """
    Fatora uma matriz definida positiva A em L * D * L^T .
    :param n: dimensão da matriz A (n x n).
    :param A: matriz simétrica e definida positiva.
    :return: Uma tupla contendo a matriz L e a lista com a diagonal D.
    """
    # Inicializa L com 1.0 na diagonal principal e 0.0 no resto.
    L = [[1.0 if i == j else 0.0 for j in range(n)] for i in range(n)]

    # Inicializa D (como é apenas uma diagonal, usamos um vetor simples para economizar memória).
    D = [0.0] * n

    # Passo 1: O laço principal que percorre as colunas[cite: 115].
    for i in range(n):
        
        # Passo 2: O vetor auxiliar 'v'[cite: 116].
        # Ele guarda as multiplicações (L * D) daquela linha para não calcularmos a mesma coisa duas vezes.
        v = [0.0] * i
        for j in range(i):
            v[j] = L[i][j] * D[j]

        # Passo 3: Calcula o elemento da diagonal de D[cite: 117].
        # Desconta do valor original de A o somatório dos elementos anteriores (L * v).
        soma1 = sum(L[i][j] * v[j] for j in range(i))
        D[i] = A[i][i] - soma1

        # Passo 4: Calcula os elementos de L abaixo da diagonal[cite: 118].
        for j in range(i + 1, n):
            soma2 = sum(L[j][k] * v[k] for k in range(i))
            # Divide pelo elemento de D que acabamos de descobrir no Passo 3.
            L[j][i] = (A[j][i] - soma2) / D[i]

    # Passo 5: Retorna as matrizes[cite: 119].
    return L, D

def solve_ldl(n, L, d, b):
    """
    Resolve o sistema linear LDL^Tx = b recebendo L como matriz 2D.
    """
    # Fase 1: Resolver Lz = b (De cima para baixo)
    z = [0.0] * n
    for i in range(n):
        soma = sum(L[i][j] * z[j] for j in range(i))
        # Note que não precisamos dividir pelo pivô de L porque L[i][i] é SEMPRE 1!
        z[i] = b[i] - soma

    # Fase 2: Resolver Dy = z (A resolução instantânea da diagonal)
    y = [0.0] * n
    for i in range(n):
        # Apenas divide os elementos pelo fator da diagonal D.
        y[i] = z[i] / d[i]

    # Fase 3: Resolver L^T x = y (De baixo para cima)
    x = [0.0] * n
    for i in range(n - 1, -1, -1):
        # O PULO DO GATO: L^T[i][j] é acessado como L[j][i].
        # Lemos as colunas de L como se fossem linhas.
        soma = sum(L[j][i] * x[j] for j in range(i + 1, n))
        x[i] = y[i] - soma

    return x
```

## Factoraçao de Cholesky


O Método da Fatoração de Cholesky é uma técnica de decomposição matricial utilizada para fatorar uma matriz simétrica e definida positiva em um produto entre uma matriz triangular inferior e sua transposta.

A fatoração é dada por
$$
A=LL^T
$$

onde:

* $L$ é uma matriz triangular inferior;
* $L^T$ é a transposta de L.

A fatoração de Cholesky é um caso particular da fatoração LU para matrizes simétricas definidas positivas. Como explora essas propriedades, ela requer aproximadamente metade das operações da decomposição LU tradicional, tornando-se um dos métodos diretos mais eficientes para essa classe de matrizes.

```python
def cholesky(n, A):
    """
    Fatora uma matriz simétrica definida positiva A no produto L * L^T.
    :param n: dimensão da matriz A (n x n).
    :param A: matriz simétrica e definida positiva.
    :return: Matriz L (triangular inferior).
    """
    # Inicializa a matriz L toda com zeros
    L = [[0.0 for _ in range(n)] for _ in range(n)]

    # Passo 1: A ponta inicial da matriz L (L[0][0])
    # Trava de segurança: se o primeiro elemento de A for negativo ou zero, 
    # a matriz não é definida positiva.
    if A[0][0] <= 0:
        print("Raiz Negativa")
        return None
    # L[0][0] é simplesmente a raiz quadrada de A[0][0]
    L[0][0] = A[0][0]**(1/2)

    # Passo 2: Preenche o resto da primeira coluna de L
    for j in range(1, n):
        L[j][0] = A[j][0] / L[0][0]

    # Passo 3: O laço principal que varre as colunas intermediárias
    for i in range(1, n - 1):
        
        # Passo 4: O elemento da diagonal principal (L[i][i])
        # Somatório dos quadrados dos elementos que já calculamos na mesma linha
        soma1 = sum(L[i][k] ** 2 for k in range(i))
        valor = A[i][i] - soma1
        
        # Nova trava de segurança para a raiz quadrada
        if valor <= 0:
            print("Raiz Negativa.")
            return None
        L[i][i] = valor ** (1/2)

        # Passo 5: Elementos abaixo da diagonal (L[j][i])
        for j in range(i + 1, n):
            # Produto interno entre a linha 'j' e a linha 'i' de L
            soma2 = sum(L[j][k] * L[i][k] for k in range(i))
            L[j][i] = (A[j][i] - soma2) / L[i][i]

    # Passo 6: O último elemento da matriz (a ponta inferior direita)
    # Segue a exata mesma lógica do Passo 4, mas isolado fora do laço
    soma3 = sum(L[n - 1][k] ** 2 for k in range(n - 1))
    valor2 = A[n - 1][n - 1] - soma3
    
    if valor2 <= 0:
        print("Raiz Negativa.")
        return None
    L[n - 1][n - 1] = valor2 ** (1/2)

    return L

def solve_cholesky(n, L, b):
    """
    Resolve o sistema linear LL^Tx = b recebendo L como matriz 2D.
    """
    # Passo 1: Resolver Ly = b (Substituição Progressiva / De cima para baixo)
    y = [0.0] * n
    for i in range(n):
        soma = sum(L[i][j] * y[j] for j in range(i))
        # Diferente do LDL^t, a diagonal de L aqui NÃO é 1, então precisamos dividir por L[i][i]
        y[i] = (b[i] - soma) / L[i][i]

    # Passo 2: Resolver L^T x = y (Substituição Regressiva / De baixo para cima)
    x = [0.0] * n
    for i in range(n - 1, -1, -1):
        # Lendo L^T ao contrário (usando L[j][i] ao invés de L[i][j])
        soma = sum(L[j][i] * x[j] for j in range(i + 1, n))
        x[i] = (y[i] - soma) / L[i][i]

    return x
```

## Factoraçao de Crout



| Método                        | Complexidade |
| ----------------------------- | ----------- |
| Eliminación Gaussiana general | (O(n^3))    |
| LU general                    | (O(n^3))    |
| **Crout tridiagonal**         | **(O(n))**  |

Como cada iteração do algoritmo especializado de Crout realiza um número constante de operações e percorre as três diagonais da matriz apenas uma vez, o custo total cresce linearmente com o tamanho do sistema. Isso o torna um método altamente eficiente para resolver problemas decorrentes da discretização de equações diferenciais, como aqueles obtidos por meio de diferenças finitas.


```python
def crout_tridiagonal(n, A, b):
    """
    Resolve o sistema linear tridiagonal usando a fatoração de Crout.
    :param n: dimensão do sistema.
    :param A: Matriz de coeficientes n x n (tridiagonal).
    :param b: vetor de termos independentes n x 1.
    :return: vetor solução x.
    """
    # Em vez de criar matrizes n x n, criamos apenas vetores de tamanho n!
    l_diag = [0.0] * n  # Guardará a diagonal principal de L
    l_sub = [0.0] * n   # Guardará a subdiagonal de L
    u_sup = [0.0] * n   # Guardará a superdiagonal de U
    z = [0.0] * n       # O vetor intermediário
    x = [0.0] * n       # O vetor da resposta final

    # ==========================================
    # FASE 1: VARREDURA DE IDA (Forward Sweep)
    # Aqui o código resolve L e z ao mesmo tempo[cite: 146].
    # ==========================================

    # Passo 1: Configuração inicial (A primeira ponta da matriz) [cite: 147-149]
    l_diag[0] = A[0][0]
    u_sup[0] = A[0][1] / l_diag[0]
    z[0] = b[0] / l_diag[0]

    # Passo 2: O laço principal que percorre o miolo da matriz [cite: 150-160]
    # Este laço faz o trabalho pesado operando apenas com os vizinhos imediatos.
    for i in range(1, n - 1):
        # Pega o elemento abaixo da diagonal [cite: 150]
        l_sub[i] = A[i][i - 1]
        
        # Calcula a nova diagonal de L descontando o passo anterior [cite: 155]
        l_diag[i] = A[i][i] - l_sub[i] * u_sup[i - 1]
        
        # Calcula a superdiagonal de U [cite: 156]
        u_sup[i] = A[i][i + 1] / l_diag[i]
        
        # Já calcula o valor de z resolvendo Lz = b "ao vivo" [cite: 159]
        z[i] = (b[i] - l_sub[i] * z[i - 1]) / l_diag[i]

    # Passo 3: O fechamento (A última ponta da matriz) [cite: 163-165]
    # Igual ao miolo, mas sem calcular U (pois não tem nada à direita da última diagonal).
    l_sub[n - 1] = A[n - 1][n - 2]
    l_diag[n - 1] = A[n - 1][n - 1] - l_sub[n - 1] * u_sup[n - 2]
    z[n - 1] = (b[n - 1] - l_sub[n - 1] * z[n - 2]) / l_diag[n - 1]

    # ==========================================
    # FASE 2: VARREDURA DE VOLTA (Backward Sweep)
    # ==========================================

    # Passo 4: O último valor de x é o último valor de z[cite: 154].
    x[n - 1] = z[n - 1]

    # Passo 5: Sobe a escada resolvendo Ux = z rapidamente [cite: 151-158].
    # Como U só tem a diagonal principal (cheia de 1s) e a superdiagonal, 
    # a equação é curtíssima.
    for i in range(n - 2, -1, -1):
        x[i] = z[i] - u_sup[i] * x[i + 1]

    # Passo 6: Retorna a solução! [cite: 161]
    return x
```

### Vantagens e Desvantagens do Método de Crout




#### Vantagens

- **Alta eficiência para matrizes tridiagonais:** O método possui complexidade computacional de **O(n)** quando aplicado a matrizes tridiagonais, tornando-o muito mais rápido do que métodos gerais de eliminação gaussiana, cuja complexidade é **O(n³)**.

- **Baixo consumo de memória:** Apenas as três diagonais da matriz precisam ser armazenadas, reduzindo significativamente o uso de memória em comparação com a manipulação da matriz completa.

- **Boa estabilidade numérica:** Para matrizes diagonalmente dominantes ou definidas positivas, o método apresenta boa estabilidade numérica sem a necessidade de pivotamento.



#### Desvantagens

- **Restrito a determinadas estruturas de matriz:** A versão otimizada do método é aplicável apenas a matrizes tridiagonais. Para matrizes densas ou com outra estrutura, é necessário utilizar a fatoração LU geral.

- **Sensível a pivôs nulos:** Caso algum elemento da diagonal principal resulte em zero durante a fatoração, o algoritmo falha devido à divisão por zero, exigindo técnicas de pivotamento.

- **Dependência das propriedades da matriz:** A estabilidade do método depende das características da matriz. Em sistemas mal condicionados, os erros numéricos podem ser amplificados.

- **Pouco vantajoso para matrizes pequenas:** Em sistemas de baixa dimensão, o ganho de desempenho em relação a métodos diretos tradicionais é pouco significativo.

## Metodo de Jacobi


O Método de Jacobi é um método iterativo utilizado para resolver sistemas lineares.

Diferentemente dos métodos diretos, como a Eliminação de Gauss, o método de Jacobi não calcula a solução exata em um único processo. Em vez disso, produz uma sequência de aproximações que convergem para a solução do sistema.

Cada nova aproximação é obtida utilizando exclusivamente os valores da iteração anterior, o que torna o algoritmo simples e facilmente paralelizável.

Para que o método convirja, é desejável que a matriz dos coeficientes seja diagonalmente dominante ou simétrica definida positiva.

## Metodo SOR


O Método SOR (Successive Over-Relaxation) é um método iterativo utilizado para resolver sistemas lineares da forma

$$
Ax=b.
$$

Ele é uma melhoria do método de Gauss-Seidel, introduzindo um fator de relaxação ω que pode acelerar significativamente a convergência.

A aproximação da solução é atualizada iterativamente utilizando os valores mais recentes já calculados. Dependendo do valor de ω, o método pode convergir mais rapidamente do que Gauss-Seidel.

Os valores do parâmetro de relaxação são interpretados da seguinte forma:

* $0<ω<1$: sub-relaxação, utilizada para tornar o método mais estável em alguns problemas.
* $ω=1$: o método torna-se exatamente o Gauss-Seidel.
* $1<ω<2$: sobre-relaxação, normalmente utilizada para acelerar a convergência.

A escolha adequada de ω depende das características da matriz do sistema.
